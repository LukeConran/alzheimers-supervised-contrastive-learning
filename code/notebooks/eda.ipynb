```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
d1 = np.load('../data/1.npz')
d2 = np.load('../data/2.npz')
print(d1.files)  # see keys: image_mr, image_parcel, etc.
```

```python
mri1 = d1['image_mr'].squeeze()
mri2 = d2['image_mr'].squeeze()
mri = np.stack([mri1, mri2])  # shape (2,128,128,128)
print(mri.shape)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12,8))

for i in range(2):
    axes[i, 0].imshow(mri[i, 64, :, :], cmap='gray')
    axes[i, 0].set_title(f'Axial (scan {i})')
    axes[i, 1].imshow(mri[i, :, 64, :], cmap='gray')
    axes[i, 1].set_title(f'Coronal (scan {i})')
    axes[i, 2].imshow(mri[i, :, :, 64], cmap='gray')
    axes[i, 2].set_title(f'Sagittal (scan {i})')

plt.tight_layout()
plt.show()
```
